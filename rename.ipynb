```python
import pandas as pd

# Load your kline CSV (it has headers, so don't use header=None)
df = pd.read_csv("ETHUSDT_1h.csv", low_memory=False)

# Convert open_time from string to datetime and set index
df["open_time"] = pd.to_datetime(df["open_time"])
df.set_index("open_time", inplace=True)

# Keep only OHLC + volume
df = df[["open","high","low","close","volume"]]

# Convert numeric columns to float
df = df.astype(float)
```

```python
import mplfinance as mpf

mpf.plot(
    df,
    type="candle",
    style="yahoo",
    volume=True,
    figsize=(12,6),
    title="ETHUSDT Hourly Candlestick Chart"
)
```
